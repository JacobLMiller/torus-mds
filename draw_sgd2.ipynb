```python
import networkx as nx 
import s_gd2
```

```python
G = nx.read_adjlist("block.txt")


I = [int(u) for u,v in G.edges()]
J = [int(v) for u,v in G.edges()]

X = s_gd2.layout_convergent(I,J)
```

```python
colors = [0] * 100 + [1] * 100 + [2] * 100 + [3] * 100 + [4] * 100
nx.draw(G,pos={u: X[i] for i, u in enumerate(G.nodes())},node_color=colors,node_size=12,width=0.4,edge_color='grey')
```

```python
import numpy as np 
np.savetxt("blcoksgd2.npy",X)
```
